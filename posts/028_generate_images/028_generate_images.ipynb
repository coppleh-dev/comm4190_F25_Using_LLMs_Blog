The prompts I will be using today (also from Gemini 2.0):

. The "Style Transfer" Prompt
Use this if you like the composition of your photo but want to change the artistic medium.

Template: "Transform the attached image into a [Style Name]. Maintain the same layout and subject, but use [Color Palette] and [Specific Brushwork/Texture]."

Example: "Transform the attached photo of the mountain range into a 19th-century oil painting. Maintain the peak placement, but use warm sunset tones and thick, visible impasto brushstrokes."

2. The "Subject Modification" Prompt
Use this if you want to keep the background/vibe but swap or add a new character or object.

Template: "Using the environment in this image as a base, replace the [Original Subject] with a [New Subject]. The new subject should match the lighting and scale of the original."

Example: "Using the forest background from this image, replace the hiker with a glowing, ethereal stag. Ensure the glow from the stag casts light on the surrounding trees from the original photo."

3. The "Atmospheric Shift" Prompt
Use this if you want the exact scene but at a different time, season, or mood.

Template: "Take this scene and reimagine it during [Time/Season/Weather]. Adjust the lighting to be [Description] and add elements like [Snow/Rain/Fog]."

Example: "Take this sunny street photography image and reimagine it as a cyberpunk noir scene at night. Add neon reflections in the puddles, heavy rain, and glowing holographic advertisements on the buildings."
