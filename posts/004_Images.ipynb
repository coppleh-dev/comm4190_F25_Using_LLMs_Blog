**<h3 style="color:purple;">The code I am using:</h3>**

```python
from openai import OpenAI
import os
import base64
import requests

from dotenv import load_dotenv
# Load API key
_ = load_dotenv()
```

```python
def encode_image(image_path):
  with open(image_path, "rb") as image_file:
    return base64.b64encode(image_file.read()).decode('utf-8')
```

```python
client = OpenAI()
```

```python
image_prompt1 = '''
            Make a marketing image for our new brand—Chinese takeout, make a picture of our main dish orange chicken. Our name is "Stir Crazy". Make it similar to marketing images other fastfood chains (ie. McDonalds use). Make the food look very appetizing. Include different sides, the title "Stir Crazy" in text, drinks, etc. Include the title "Stir Crazy." Make it look more like savory chinese food. Make all content in english.'''
```

```python
response = client.images.generate(
  model="dall-e-3",
  prompt=image_prompt1,
  size="1024x1024",
  quality="standard",
  n=1,
)

image_url = response.data[0].url
```

```python
generated_image = requests.get(image_url)

with open('idea_one_img.png', 'wb') as out:
    out.write(generated_image.content)
```
