Linked-In Cofounder Reid Hoffman in Superagency argues that AI will be a transformative tool that allows us to reach this state called Superagency—a new level of human capability and progress. However, as you read the book from discussing productivity to "new job creation" the authors appear out of touch. Will AI be transformative for ALL or a productivity machine and proft-making tool for the elite few who can use it to their own advantage? Today, I am going to unpack this. 

**The Regulatory Moat**

Hoffman and Beato argue to regulate AI as issues arise, rather than slowing down innovation for regulation (especially in a free country such as the U.S.). This process of gradually releasing and testing AI through emerging it into components of broader society is coined as iterative deployment. They also argue every technology (ie. the car) have costs associated (ie. car accidents) but the benefits out weight these costs. For AI, the authors argue the same emphasis no freedom to explore and study must be implemented.

**Is Individuality the Key to Success?**

Hoffman and Beato argue that poverty and inequality can be solved at the individual level because AI will ease access to information, tutoring, therapy, etc. However, Hoffman and Beato fail to account how AI can also erode deep, critical thinking skills. AI can offload critical thinking in users, making individuals less capable of independent problem-solving. Although AI may help in completing a worksheet or writing an essay, the fundamental skills to thinking and problem solving are at risk with LLMs.

**Pathos**

At the end of the day, Superagency is written by a billionaire insider who sees AI as a positive-sum game. While Hoffman may share his excitement over AI's potential to raise all wages, other findings share that the top 1% will benefit the most from AI.
