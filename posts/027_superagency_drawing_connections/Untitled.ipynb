Hoffman and Beato suggest that AI engineers try to "maintain strict separation between the data they use to train their models and the data they use to test their models to prevent data leaks." If an LLM is trained on too large a data pool, of course it can match certain pieces of information but this does not make it generative, thus creating a need for separation.

What Hoffman and Beato refer to as data leakage occurs when the training set and test set have overlapping points.
