In Greg Beato and Reid Hoffman's book *Superagency* they use GPS (Global Positioning System) as a central metaphor for AI.

Hoffman and Beato argue AI is changing the Information Landscape just as GPS did, as AI helps users navigate a "destination"(ie. a prompt/task) and delivers them the steps on HOW to get there and the final destination. 

Hoffman and Beato consistently describe how we are living in an era of "Big Knowledge" where this is a vast sea of information that is too large for a single human to navigate. Like a GPS in the vast map of earth, AI processes billions of data points to locate and pinpoint where to look in the vast web of human knowledge.

Further, Hoffman and Beato see AI as a dynamic system that can help you pivot (like a GPS would for traffic), helping you re-route and redirect on where to look and utilize the vast web of data.

**My Opinion**:

My bigges critique of this metaphor is that GPS relies on an objective, phsyical reality while an LLM can hallucinate data points that have never existed. AI operates on likelihood not truth, and it poses a risk of with, too much dependency, losing the ability to think and find important information.

A GPS is helpful because we don't need maps and compasses to survive in society...but we do still need our brains. 

If AI is our new GPS for, well, life (readings, writing, math, thinking) then what do we become? I think there is something beautiful in the quest for learning (although difficult and challenging at times). But there is something beautiful to be said to sit without technology and absorb difficult content and concepts and push your brain and yourself. I think we are really losing this with AI. Hoffman and Beato talk about all the things AI can replace, essentially arguing it can become a crutch for so many different activities. But, at one point, are we allowing AI to replace or mechanize the human experience? Critical thinking skills and uncovering solutions without a playbook next to you are vital experiences that grow generations of leaders and innovators. I think we may be losing that to AI. If we are exposed to the solution to a problem within 5 seconds, we lose the innate, human struggle of coming to that answer.

Maybe thinking does not need a GPS, but rather the skills that get you to your destination. 

Hoffman and Beato are right—we don't need maps anymore, but we do need our brains. 

Let's not allow AI to replace those, ok?
