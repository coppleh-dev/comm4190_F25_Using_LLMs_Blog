I've always wondered if I have a look a like or a doppleganger out there. Better yet, do I have a celebrity look alike. So, today, I am asking Gemini (by uploading pictures of myself) if there are any celebrities who look like me!!

**Reflection: Hahaha even though I do not think I look like these celebrities it was still a fun thought experiment...a little under impressed because these are 3 key major celebrities, but that's okay! Next I want to explore the mechanics of how this works!**

**Analysis**: When you upload an image it is broken down into tokens, just like a string of text would be. A neural network processes these tokens and an image is converted into vector embeddings. The LLM measures how close embedddings from one picture are to another and my vector embedding "coordinates" are near another celebrities then that image will have a match. However, because this is based on probability rather than truth the model can be biased by lighting and angle, or if you pose like an image of a certain celebrity the coordinates may be closer. Further, the model might suggest a famous person because there's more data on them. 


Also, whne I asked ChatGPT, the LLM responded that they cannot do comparisons, which I found interesting. I'm sure there's some data privacy component involved.
