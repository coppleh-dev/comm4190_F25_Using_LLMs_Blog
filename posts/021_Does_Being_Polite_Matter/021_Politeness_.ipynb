I don't know about you, but I am pretty darn nice to my AI. "Please write me..", "Could you brainstorm..", followed by a thank you or a no, but please etc etc. So, why are polite to a series of mathematical equations..? Does it matter if we're polite? Do you feel guilt towards an LLM after being rude? Well, today I want to unpack all of this.

**The Mirror Effect: AI as a Social Reflection**

One of the most compelling reasons as to why we are kind to our LLMs is that we are social creatures, thus when something (even a math equation) responds to us in a human-like way, we respond back. The majority of the public also don't know how exactly AI works, leading to a tendency to anthropomorphize your LLM or even grow an attatchement to the model. 

**Does it matter?**

The short answer is...no! (Confirmed from my best buddy Gemini Flash 2.0).




So... the LLM does not "care" if you are rude, neither does it--from a utilitarian perspective--make it better or more efficient. If we include "please" in our prompts it simply becomes another **token** for the LLM to input which may make its response more polite, but not the quality of the answer it will give. 

**My Opinion: Let's Still Be Polite!**

I believe that WE as social, human creatures beneft from being polite to our AI. Maintaining decorum and cordiality (yes even with a probabilistic mathematical equation chatbot) helps us reinforce our own values and shape how we, as real humans, behave.
