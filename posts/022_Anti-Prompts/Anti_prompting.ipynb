As I've discussed before LLMs aren't thinkers..they are basically just trying to guess the next best word based on training and test data using probability. So, today, I want to break my LLM to better understand how they predict next words and to make an AI "creative." 

Today, I plan on putting constraints on the LLM (Gemini 2.0 Flash) to make it react and adapt to counter-intuitive word predeiction.I prompt Gemini to create a list to feed itself for counterintuitive prompting:


The Forbidden Word List: Ask for a travel guide to Paris but forbid the use of the words "Eiffel," "Croissant," "Romantic," or "Art." Now the AI can't rely on its most probable tokens. It has to dig deeper into its training data to find something new.

The Incongruent Persona: Instead of "Act as a chef," try "Act as a chef who secretly hates food and wishes they were a structural engineer." The resulting recipe will be bizarre, fascinating, and likely something no human has ever written before.

The "Lying" Prompt: Tell the AI something factually incorrect as a premise. "Write a history of the Great Balloon War of 1842." Because the AI is trying to remain helpful, it will "hallucinate" a logic to support your lie, often creating world-building details that are far more creative than a standard fantasy prompt.

1. The "Forbidden Word" Constraint
The Goal: To get a travel description that isn't full of clichés.

Prompt: "Describe the experience of visiting a bustling night market in Tokyo. Constraint: You are strictly forbidden from using the words: Neon, Sushi, Busy, Tradition, Culture, Crowded, or Delicious. Focus instead on the sounds of the machinery and the geometry of the stalls."

2. The Incongruent Persona
The Goal: To get a unique perspective on a mundane topic.

Prompt: "Explain the process of brewing a perfect cup of coffee, but act as a grumpy medieval fortress commander who views the kitchen equipment as siege engines and the coffee beans as enemy spies."

3. The "Lying" Premise (Counter-Factual History)
The Goal: To spark world-building ideas for a story or game.

Prompt: "Write a short encyclopedia entry about the 'Great Library of 1994,' a physical building that housed the entire early internet before it was destroyed by a digital virus that became sentient. Note: Treat this as a historical fact and describe the architecture of the 'Gmail Wing.'"

4. The "Negative Space" Design
The Goal: To get a product idea that doesn't follow standard trends.

Prompt: "Pitch me a new smartphone design for the year 2030. Constraint: The device cannot have a screen, it cannot use voice commands, and it cannot be made of glass or metal. How does the user interact with it?"

5. The "Logic Leap" (Anti-Context)
The Goal: To find a "statistical outlier" in the AI's vocabulary.

Prompt: "Write a poetic description of a sunset, but use only terminology from quantum physics and automotive repair. Avoid all words related to color (like red, orange, or gold)."
