I've been thinking a lot about my good old fashioned Google Translating Days and, how, if I ever used Google Translate my Spanish teachers would know **immediamente* because GT used stiff, non-conversational language. So, I want to test if Gemini Flash 2.0 can translate Spanish intuitively and, of course, as a non-native Speaker I shall have my native spanish-speaking Boyfriend review. 

 I'm going to start out with a simple prompt to translate englsih to spanish:
*"Translate the following English text into natural, conversational Spanish (like two friends would speak). Do not use overly formal language or literal word-for-word translations if a more common, idiomatic phrase exists.*

**Prompt #1**

*"Translate the following English text into natural, conversational Spanish (like two friends would speak). Do not use overly formal language or literal word-for-word translations if a more common, idiomatic phrase exists: im so bored in class today, all I want to do is lie in my bed and watch white lotus.*

**Output**

Here is the translation of that English text into natural, conversational Spanish:

"¡Qué aburrido estoy en clase hoy! Lo único que quiero es tirarme en la cama a ver The White Lotus."


**Reflection**

My boyfriend reviewed this as a "little stiff," arguing it translated word for word. I'm a bit disappointed becauase I had hope that Gemini Flash would be better at imitating linguistic variabilty across languages. Next, I am going to try a more specific prompt.

**Prompt #2**

"Your task is to act as a professional translator specializing in modern, spoken Spanish. Translate the following English text into Spanish, focusing on making the translation sound idiomatic, relaxed, and conversational. Use common Spanish contractions, slang (if appropriate and widely understood), and natural flow. The target audience is peers in an informal setting. Translate:  im so bored in class today, all I want to do is lie in my bed and watch white lotus.

**Output**

Here is the translation of that English text into natural, conversational Spanish:

"¡Qué aburrimiento tengo en clase hoy! Solo quiero tumbarme en la cama y ponerme a ver The White Lotus."


**Reflection**

Yikes..he is still saying it is bad, although the second half of the sentence is an improvement. Let's try our last prompt to get a more accurate interpretation. 

**Prompt #3**

"Hey, can you translate this for me? I need it to sound totally normal and casual, like something a native speaker would actually say in real life, not something from a textbook. Use all the usual contractions and make sure it flows well.

**Output**

"Qué pereza/aburrimiento tengo hoy en clase. Lo único que quiero es tirarme en la cama a ver The White Lotus."




**Reflection**

Now, in Spain this would be accurate, but anywhere else...nope. This makes me wonder how Gemini is trained to translate languages. I looked it up and Gemini uses a Transformer Architecture, Reinforcement Learning, and and Supervised Fine-Tuning. This is interesting to me because I am taking a Machine Learning class on multimodal systems and transformer architecture as a layered architecture. But, overall, do better Gemini when translating!
